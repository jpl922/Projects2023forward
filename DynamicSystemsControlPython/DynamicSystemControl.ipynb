```python
import sympy
sympy.init_printing()
```

```python
D,d,H,h = sympy.symbols('D,d,H,h', real = True)
```

```python
R = D/2
r = d/2
Hprime = sympy.sqrt(H**2 - (R-r)**2) # pythagoras; hate geometry
```

```python
radius = r +h/Hprime*(R-r)
```

```python
A = sympy.pi*radius**2
```

```python
V = sympy.integrate(A, (h,0,h))
```

```python
V
```

```python
print(V)
```

```python
Vsymb = sympy.symbols('V', real = True)
```

```python
hV = sympy.solve(Vsymb -V, h)
```
