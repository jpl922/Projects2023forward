```python
credits = [8,16,8,8,16]
marks = [75, 60, 60, 75, 45]
```

```python
weightedsum = 0
creditsum = 0 
for i in range(len(credits)):
    weightedsum+= credits[i]*marks[i]
    creditsum += credits[i]

avg = weightedsum/creditsum
print(avg)
```

```python
weightedsum =0
creditsum =0
for credit, mark in zip(credits,marks): # zip iterator which groups its inputs into tuples to expand in the loop
    weightedsum += credit*mark
    creditsum += credit
avg = weightedsum/creditsum
print(avg)
```

```python
def f(x):
    return x**2 -3
```

```python
import scipy
```

```python
import scipy.optimize
```

```python
scipy.optimize.fsolve(f,2) # requires function as the argument and takes bunch of code
```

```python
scipy.optimize.fsolve(lambda x: x**2 -3, 2) # quick anonymous function solves same as f,2 but one line
```
