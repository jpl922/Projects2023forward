what are the independence requirements? Schedule risk does not feel like all events are truly independent, but probably enough? 

```python
import numpy as np 
sims = 1000000 
A = np.random.uniform(1,5, sims) # task A is uniform distribution between 1 and 5; takes 1 million points
B = np.random.uniform(2,6, sims) # task B is uniform distribution between 2 and 6; takes 1 millions points
duration = A+B # duration is the sume of the 1 millions A,B points resulting in 1
```

```python
import matplotlib.pyplot as plt # Plotting historication of the 1000000 monte carlo simulation; red line is 9 hr limit; 
plt.figure(figsize = (3, 1.5))
plt.hist(duration, density = True)
plt.axvline(9, color = 'r')
plt.show()
print((duration>9).sum()/sims) # print percentage > 9 hours 
```
